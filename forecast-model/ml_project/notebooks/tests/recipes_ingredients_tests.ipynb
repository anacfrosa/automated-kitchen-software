```python
import pandas as pd

pd.set_option('display.max_columns', 23)
df_all_food_com = pd.read_csv('csv_files/food_com_db_w_units.csv') ## dataset with a register of food recipes 

# Create a new dataframe associating each recipe to a group of ingredients with his measures, units and servings
# Servings - For how many people is the recipe
df_recipes_with_ings = df_all_food_com.groupby('title').agg({'ingredient': lambda x: list(x), 'measure': lambda x: list(x), 
                                                             'unit': lambda x: list(x), 'servings': lambda x: list(x)}).reset_index()
df_recipes_with_ings.columns = ['title', 'ingredients', 'measures', 'units', 'servings']
```

```python
df_recipes_with_ings
```

```python
# Convert all the title letters to lower case 
df_recipes_with_ings['title'] = df_recipes_with_ings['title'].str.lower()
```

```python
df_recipes_with_ings
```

```python
# Check if there is duplicate recipes
duplicate_titles = df_recipes_with_ings[df_recipes_with_ings.duplicated(subset=['title'], keep=False)]
#duplicate_titles
```

```python
# Remove all the duplicate recipes
df_recipes_with_ings = df_recipes_with_ings.drop_duplicates(subset=['title'])
```

```python
df_recipes_with_ings
```

```python
# Search through the exact word 
search_name = 'gobi manchurian'
found_recipes = df_recipes_with_ings[df_recipes_with_ings['title']==search_name]
```

```python
found_recipes
```

```python
# Search through the exact word with isin function
search_names = ['gobi manchurian']
found_recipes = df_recipes_with_ings[df_recipes_with_ings['title'].isin(search_names)]
```

```python
found_recipes
```

```python
search_names = ['sarson da saag']
# Filtering out the rows containing the found recipes
found_recipes = df_recipes_with_ings[df_recipes_with_ings['title'].str.contains('|'.join(search_names))]
```

```python
found_recipes
```

```python
# Dropping the found recipes from df_recipes_with_ings
df_recipes_with_ings = df_recipes_with_ings.drop(found_recipes.index)
```
